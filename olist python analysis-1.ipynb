```python
import pandas as pd

orders = pd.read_csv('olist_orders_dataset.csv')
orders.head()
```

```python
order_items = pd.read_csv('olist_order_items_dataset.csv')
products = pd.read_csv('olist_products_dataset.csv')
sellers = pd.read_csv('olist_sellers_dataset.csv')
customers = pd.read_csv('olist_customers_dataset.csv')
reviews = pd.read_csv('olist_order_reviews_dataset.csv')

print(order_items.shape)
print(products.shape)
print(sellers.shape)
print(customers.shape)
print(reviews.shape)
```

```python
df = order_items.merge(orders, on='order_id', how='left')
df = df.merge(products, on='product_id', how='left')
df = df.merge(sellers, on='seller_id', how='left')
df = df.merge(customers, on='customer_id', how='left')
df = df.merge(reviews, on='order_id', how='left')

print(df.shape)
df.head()
```

```python
df['order_estimated_delivery_date'] = pd.to_datetime(df['order_estimated_delivery_date'])
df['order_delivered_customer_date'] = pd.to_datetime(df['order_delivered_customer_date'])

df['delay_days'] = (df['order_delivered_customer_date'] - df['order_estimated_delivery_date']).dt.days

df['delay_days'].describe()
```

```python
df_delivered = df.dropna(subset=['order_delivered_customer_date']).copy()

df_delivered['is_late'] = df_delivered['delay_days'] > 0

print(df_delivered.shape)
print(df_delivered['is_late'].value_counts())
print(df_delivered['is_late'].mean())
```

```python
seller_stats = df_delivered.groupby('seller_id').agg(
    total_orders=('order_id', 'count'),
    late_orders=('is_late', 'sum'),
    avg_review_score=('review_score', 'mean'),
    avg_delay_days=('delay_days', 'mean')
).reset_index()

seller_stats['late_rate'] = seller_stats['late_orders'] / seller_stats['total_orders']

seller_stats_filtered = seller_stats[seller_stats['total_orders'] >= 20].copy()

print(seller_stats.shape)
print(seller_stats_filtered.shape)
seller_stats_filtered.sort_values('late_rate', ascending=False).head(10)
```

```python
seller_orders_with_context = df_delivered.groupby('seller_id').agg(
    seller_state=('seller_state', 'first'),
).reset_index()

seller_stats_filtered = seller_stats_filtered.merge(seller_orders_with_context, on='seller_id', how='left')

state_baseline = df_delivered.groupby('customer_state')['is_late'].mean().rename('state_avg_late_rate')

df_delivered = df_delivered.merge(state_baseline, on='customer_state', how='left')

print(seller_stats_filtered[['seller_id','seller_state','late_rate']].head())
print(df_delivered[['customer_state','state_avg_late_rate']].drop_duplicates().sort_values('state_avg_late_rate', ascending=False).head(10))
```

```python
seller_expected = df_delivered.groupby('seller_id')['state_avg_late_rate'].mean().rename('expected_late_rate').reset_index()

seller_stats_filtered = seller_stats_filtered.merge(seller_expected, on='seller_id', how='left')

seller_stats_filtered['late_rate_vs_expected'] = seller_stats_filtered['late_rate'] - seller_stats_filtered['expected_late_rate']

seller_stats_filtered.sort_values('late_rate_vs_expected', ascending=False).head(10)[
    ['seller_id','seller_state','total_orders','late_rate','expected_late_rate','late_rate_vs_expected','avg_review_score']
]
```

```python
result = seller_stats_filtered.sort_values('late_rate_vs_expected', ascending=False).head(10)[
    ['seller_id','seller_state','total_orders','late_rate','expected_late_rate','late_rate_vs_expected']
]
pd.set_option('display.max_columns', None)
result
```

```python
category_baseline = df_delivered.groupby('product_category_name')['is_late'].mean().rename('category_avg_late_rate')

df_delivered = df_delivered.merge(category_baseline, on='product_category_name', how='left')

seller_category_expected = df_delivered.groupby('seller_id')['category_avg_late_rate'].mean().rename('expected_late_rate_category').reset_index()

seller_stats_filtered = seller_stats_filtered.merge(seller_category_expected, on='seller_id', how='left')

seller_stats_filtered['combined_expected'] = (seller_stats_filtered['expected_late_rate'] + seller_stats_filtered['expected_late_rate_category']) / 2

seller_stats_filtered['late_rate_vs_combined_expected'] = seller_stats_filtered['late_rate'] - seller_stats_filtered['combined_expected']

pd.set_option('display.max_columns', None)
seller_stats_filtered.sort_values('late_rate_vs_combined_expected', ascending=False).head(10)[
    ['seller_id','seller_state','total_orders','late_rate','expected_late_rate','expected_late_rate_category','combined_expected','late_rate_vs_combined_expected','avg_review_score']
]
```

```python
avg_order_value = df_final['price'].mean()
print(f"Average item price: R${avg_order_value:.2f}")

late_orders_df = df_final[df_final['is_late'] == True]
print(f"Total late orders (row-level): {len(late_orders_df)}")
print(f"Total revenue tied to late orders: R${late_orders_df['price'].sum():,.2f}")
```

```python
late_review_avg = df_final[df_final['is_late'] == True]['review_score'].mean()
ontime_review_avg = df_final[df_final['is_late'] == False]['review_score'].mean()

print(f"Avg review score - late orders: {late_review_avg:.2f}")
print(f"Avg review score - on-time orders: {ontime_review_avg:.2f}")

bad_review_late = df_final[(df_final['is_late'] == True) & (df_final['review_score'] <= 2)]
bad_review_rate_late = len(bad_review_late) / len(late_orders_df)

bad_review_ontime = df_final[(df_final['is_late'] == False) & (df_final['review_score'] <= 2)]
bad_review_rate_ontime = len(bad_review_ontime) / len(df_final[df_final['is_late'] == False])

print(f"Bad review rate (score<=2) when late: {bad_review_rate_late:.1%}")
print(f"Bad review rate (score<=2) when on-time: {bad_review_rate_ontime:.1%}")
```

```python
excess_bad_review_rate = bad_review_rate_late - bad_review_rate_ontime

revenue_at_risk = late_orders_df['price'].sum() * excess_bad_review_rate

print(f"Excess bad-review rate caused by lateness: {excess_bad_review_rate:.1%}")
print(f"Estimated revenue at risk from late deliveries: R${revenue_at_risk:,.2f}")
```

```python
top_offenders = seller_stats_filtered.sort_values('late_rate_vs_combined_expected', ascending=False).head(20)['seller_id']

top_offender_late_orders = df_final[(df_final['seller_id'].isin(top_offenders)) & (df_final['is_late'] == True)]

revenue_at_risk_top20 = top_offender_late_orders['price'].sum() * excess_bad_review_rate

print(f"Late orders from top 20 offending sellers: {len(top_offender_late_orders)}")
print(f"Revenue at risk from top 20 offending sellers alone: R${revenue_at_risk_top20:,.2f}")
print(f"That's {revenue_at_risk_top20/revenue_at_risk:.1%} of total revenue-at-risk, from just 20 out of {len(seller_stats_filtered)} qualifying sellers")
```

```python
for n in [20, 50, 100, 200]:
    top_n = seller_stats_filtered.sort_values('late_rate_vs_combined_expected', ascending=False).head(n)['seller_id']
    late_n = df_final[(df_final['seller_id'].isin(top_n)) & (df_final['is_late'] == True)]
    risk_n = late_n['price'].sum() * excess_bad_review_rate
    print(f"Top {n} sellers ({n/len(seller_stats_filtered):.1%} of pool): {risk_n/revenue_at_risk:.1%} of total risk")
```

```python
seller_late_revenue = df_final[df_final['is_late'] == True].groupby('seller_id')['price'].sum().rename('late_order_revenue').reset_index()

seller_stats_filtered = seller_stats_filtered.merge(seller_late_revenue, on='seller_id', how='left')

seller_stats_filtered['revenue_at_risk'] = seller_stats_filtered['late_order_revenue'] * excess_bad_review_rate

seller_stats_filtered.to_csv('seller_accountability_summary.csv', index=False)
print("Updated and re-saved")

seller_stats_filtered.sort_values('revenue_at_risk', ascending=False).head(10)[
    ['seller_id','seller_state','total_orders','late_rate_vs_combined_expected','late_order_revenue','revenue_at_risk']
]
```
